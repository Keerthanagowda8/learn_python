# Bubble Sort

- its a simple sorting algorithm that repeatedly steps through the list, compares adjacent elements and swaps them if they are in the wrong order. The pass through the list is repeated until the list is sorted.

- Time Complexity: $O(n^2)$ Average & Worst Case | $O(n)$ Best Case (if already sorted)
- Space Complexity: $O(1)$ (In-place)


## How it Works (Example)

Let's sort [5, 1, 4, 2]

Pass 1:

Compare 5 and 1 $\rightarrow$ Swap $\rightarrow$ [1, 5, 4, 2]

Compare 5 and 4 $\rightarrow$ Swap $\rightarrow$ [1, 4, 5, 2]

Compare 5 and 2 $\rightarrow$ Swap $\rightarrow$ [1, 4, 2, 5] (5 is now in its final position!)

Pass 2:

Compare 1 and 4 $\rightarrow$ No swap

Compare 4 and 2 $\rightarrow$ Swap $\rightarrow$ [1, 2, 4, 5] (4 is now in its final position!)

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
        print(arr)  # Print the array after each pass
    return arr


arr = [5, 1, 4, 2]
sorted_arr = bubble_sort(arr)
print("Sorted array:", sorted_arr)
```

note:
 
    here we can see, in the 2nd pass only we got the sorted array. but it again goes through the entire array. we need to optimize by adding a flag to check if any swaps were made in the current pass. If no swaps were made, the array is already sorted and we can break early.

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False  # Flag to check if any swaps were made
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True  # Set the flag to True if a swap was made
        if not swapped:  # If no swaps were made, the array is already sorted
            break
        print(arr)  # Print the array after each pass
    return arr


arr = [5, 1, 4, 2]
sorted_arr = bubble_sort(arr)
print("Sorted array:", sorted_arr)
```

# Quck Sort

- it is a advanced sorting technique which helps us to sort the elements with the help of divide and conquer role
- Quick Sort picks an element as a pivot and partitions the array around it, moving all smaller elements to its left and all larger elements to its right. It then recursively applies the same logic to the left and right sub-arrays


Time Complexity: $O(n \log n)$ Average | $O(n^2)$ Worst Case (if pivot selection is poor)

Space Complexity: $O(\log n)$ (Logarithmic stack space)


## How it Works (Example)
Let's sort [10, 80, 30, 90, 40] using 40 as the pivot:

Compare items to 40. 10 and 30 are smaller; 80 and 90 are larger.

Partition: Rearrange so smaller elements come before 40 and larger elements come after $\rightarrow$ [10, 30, 40, 90, 80].

Recursively sort the left side [10, 30] and right side [90, 80]


```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2] # Choosing middle element as pivot
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        print(f"Pivot: {pivot}, Left: {left}, Middle: {middle}, Right: {right}")  # Print the pivot and partitions
        return quick_sort(left) + middle + quick_sort(right)


arr = [11,22,45,2, 1, 4, 3]
quick_sort(arr)
```

# Merge Sort

- it is a advanced sorting technique which helps us to sort the elements with the help of divide and conquer role
- in this first need to divide the elements into two equals and continue the process until list cannot be divided into two equals.
- when dividing is stops start merging the list by comparing left and right elements

        ---------------------------------
        | 4 | 8 | 1 | 6 | 7 | 3 | 5 | 9 |
        ---------------------------------
                /               \
        ----------------     -----------------
        | 4 | 8 | 1 | 6 |    | 7 | 3 | 5 | 9 | 
        ----------------     -----------------
            /         \              /         \
        --------     ---------     ---------     ---------
        | 4 | 8 |    | 1 | 6 |     | 7 | 3 |     | 5 | 9 |
        ---------    ---------     ---------     ---------
            /   \          /   \           /  \           /      \
        -----   -----    -----   -----   -----   -----    -----   -----
        | 4 |   | 8 |    | 1 |   | 6 |   | 7 |   | 3 |    | 5 |   | 9 |
        -----   -----    -----   -----   -----   -----    -----   -----

        


```python
def merge_sort(l1):
    if len(l1) > 1:
        mid = len(l1) // 2
        left_half = l1[:mid]
        right_half = l1[mid:]

        print("left half: ",left_half)
        print("right half: ", right_half)
        print(mid)

        merge_sort(left_half)
        merge_sort(right_half)

        # start megring the list by comparing

        i = 0   # points the left list 
        j = 0   # points the right list 
        k = 0   # points the original list 

        # iam using while loop because we don't know the no of iteration
        while i < len(left_half) and j < len(right_half):

            if left_half[i] < right_half[j]:
                l1[k] = left_half[i]
                i += 1

            else:
                l1[k] = right_half[j]
                j += 1

            k += 1

        # check any element is remaining in both list (like some element are same or only one side have element then that won't check in the first while loop)

        while i < len(left_half):
            l1[k] = left_half[i]
            i += 1
            k += 1

        while j < len(right_half):
            l1[k] = right_half[j]
            j += 1
            k += 1

    print(l1)


l1 = [4,8,1,6,7,3,5,9]

print("before sorting: ",l1)
print("after sorting")
merge_sort(l1)
```

# Insertion Sort

- it is a sorting algorithm which is efficient for both huge amount of elements and limited number of elements 
- it builds the final sorted array one item at a time.
- it takes each element from the unsorted portion and shifts it into its correct position within the already sorted portion


Time Complexity: $O(n^2)$ Average & Worst Case | $O(n)$ Best Case

Space Complexity: $O(1)$ (In-place)


## algorithm

- consider given list into two parts like sorted and unsorted list
- in that first element must be in the sorted list and remaining all the element must be in unsorted list

          sorted list           unsorted list
               \             /  
               |           |   
          ------------------------------------------
          | 8  |  3  |  1  |  2  |  7  |  6  |  5  |
          ------------------------------------------
               |

- start the iteration by comparing both the list elements
- take first element from unsorted list and compare with last element in sorted list like sorted list element is greater than unsorted list
- if it is true swap the elements

- the above process should continue till unsorted list element becomes empty


## How it Works (Example)
Let's sort [4, 3, 2, 10]:

Start at index 1 (value 3). Compare with 4 $\rightarrow$ Shift 4 right $\rightarrow$ Insert 3 $\rightarrow$ [3, 4, 2, 10]

Move to index 2 (value 2). Compare with 4 and 3 $\rightarrow$ Shift both right $\rightarrow$ Insert 2 $\rightarrow$ [2, 3, 4, 10]

Move to index 3 (value 10). It's already larger than 4 $\rightarrow$ Stays put $\rightarrow$ [2, 3, 4, 10]

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        # print(arr)
        arr[j + 1] = key
        print(arr)  # Print the array after each pass
    return arr
    

arr = [1,2,3]
print(insertion_sort(arr))
```
