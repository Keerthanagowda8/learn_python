# QUEUE

- it is a linear dynamic data structure works with the principle of FIFO (First In First Out)
- we need to create queue data structure with the help of the list
- in this inserstion of elements is happening at one end called rear end and deletion of elements is happening at another end called front end
-  in this we have 3 types
1. linear queue
    - in this elements are arranged in a sequential manner with or without limited size
    
            |  10  |  20  |  30  |  40  |
                0      1      2      3

2. circular queue
    - in this the elements are arranged in sequential and circular mannar
    - i should be created with the help of limited size
    - it is also called as ring buffer
    - in this we can reuse the empty space created by deletion of elements

                |  10  |  20  |  30  |  40  |
                    0      1      2      3
                \                          /
                 \                        /
                  \----------------------/

3. priority queue
    - it will work based on the priority of a given elements

            |                                                  |
            | [ 'task A', 3 ] [ 'task B', 1 ] [ 'task C', 2 ]  |
            |                                                  |


            [ 'task B', 1 ]
            [ 'task C', 2 ]
            [ 'task A', 3 ]

- common operations on queue

1. enqueue(): To add elements
2. dequeue(): To remove elements
3. peek(): To display first inserted element
4. Display(): To display all element
5. isEmpty(): To check whether the queue is empty or not
6. isFull(): To check whether the queue is full or not

```python
class QUEUE:
    def __init__(self):
        self.queue = []

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        self.queue.pop(0)

    def peek(self):
        return self.queue[0]
    
    def display(self):
        print(self.queue)


q = QUEUE()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

q.display()

q.dequeue()
q.display()

q.peek()
```

# circular queue

- it is a unique queue datastructure works with the principal of first in first out
- in this the queue must be created with limited size and all the cell values must be none initially
- in this the elements need to arrange or store in a circular manner, need to have two variables front and rear
and initialy it should contains value -1
- front will be taken care of deletion of elements from circular queue and rear is responsible to insert the element to circular queue

## Algorithmic steps

1. IsFull():
    - check front is pointing to zeroth index and rear is reached maximum size in cq
    if it is return the boolean value True

            |  A  |  B  |  C  |  D  | -iSfULL()-> True
            0     1     2     3
            ^                 ^
            |                 |
            FRONT              REAR

    - check front is pointing to one index to the rear if it is return TRUE

            |  F  |  G  |  C  |  D  |  -iSfULL()-> True
            0     1     2     3
                    ^     ^
                    |     |
                    FRONT  REAR

    - if the none of the condition True return False in else point

2. IsEmpty():
    - check if both front and rear is pointing to -1 then return boolean value True
    - or return boolean value False in else part
    
3. enqueue():
    - check circular queue is full or not by calling IsFull method, if it returns True diaplay a message
    - check if rear is pointing to maxsize bring back to zeroth position or increment the rear position
    - check front is pointing to -1 or not before inserting, if it is True Print the front to zeroth postion
    - insert the element to circular queue where rear is pointing.

4. dequeue():
    - check circular queue is empty or not by calling isEmpty method, if it is true display a message
    - if it returns false in the else part check if both front and rear is pointing to the same index make it as -1
    
        
            |  F  |  None  |  None  |  None  |  -iSfULL()-> True
            0       1       2        3
            ^     
            |     
            FRONT,REAR

    - check if pointing to maxsize bring back to 0th element



```python
class CircularQueue:
    def __init__(self,maxsize):
        self.cq = [None] * maxsize
        self.maxsize = maxsize
        self.rear = -1
        self.front = -1
    
    def display(self):
        print(self.cq)
        print("front-> ",self.front)
        print("rear-> ",self.rear)

    def IsFull(self):
        if self.front == 0 and self.rear + 1 == self.maxsize:
            return True
        elif self.rear + 1 == self.front:
            return True
        else:
            return False
        
    def IsEmpty(self):
        if self.front == -1 and self.rear == -1:
            return True
        else:
            return False
        
    def enqueue(self,val):
        if self.IsFull():
            print("Cricular queue is Full")
        else:
            if self.rear + 1 == self.maxsize:
                self.rear = 0
            else:
                self.rear += 1
                if self.front == -1:
                    self.front = 0
            self.cq[self.rear] = val

    def dequeue(self):
        if self.IsEmpty():
            print("circular queue is empty")
        else:
            start = self.front
            if self.front == self.rear:
                self.front = -1
                self.rear = -1
            elif self.front + 1 == self.maxsize:
                self.front = 0
            else:
                self.front += 1
            self.cq[start] = None

    def peek_ele(self):
        if self.IsEmpty():
            print("circular queue is empty")
        else:
            print("peek element: ", self.cq[self.front])

            


        
c1 = CircularQueue(5)

c1.display()
     
c1.enqueue('A')
c1.enqueue('B')
c1.enqueue('C')
c1.enqueue('E')
c1.display()
c1.enqueue('F')
c1.enqueue('G')
c1.display()

c1.dequeue()
c1.display()
c1.dequeue()
c1.display()
c1.dequeue()
c1.display()
c1.enqueue('G')
c1.display()
c1.enqueue('H')
c1.display()
c1.enqueue('I')
c1.display()
c1.peek_ele()
c1.enqueue('J')
c1.display()
c1.peek_ele()
c1.dequeue()
c1.display()
c1.peek_ele()

```
