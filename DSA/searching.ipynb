# Searching algo

1. Linear search

- it will work for both homo and hetero and sorted and unsorted elements also


```python
def linear_search(l1,val):
    for i in range(len(l1)):
        if l1[i] == val:
            return f"value found at {i}"
    else:
        return -1


l1 = [10,12,3,41,52]
val = int(input("enter the value to search: "))
print(linear_search(l1, val))
```

2. binary search

- it will only work in homo and sorted list
- it is a most efficient seraching technique in dsa 
- this algorithm will work only for sorted list elements
- and it will work only for integers

## algorithm

1. start the iteration by checking start index should be always less than end index

            -----------------------------------------------
            | 8  |  10  |  14  |  22  |  27  |  36  | 55  |
            -----------------------------------------------
              |                                    |
            start          start <= end           end      

2. find the mid index with the help of start and end index

    mid = (start + end) // 2

3. check mid value is equal to search value or not if it is true return the mid
4. if above condition is false check search value is greater than mid value or not, if it is true assign the start to mid+1 postion

    val > mid value
        start = mid+1

5. check if search value is less than mid value, if it is true assign end to mid-1 position

    val < mid value
        end = mid - 1

6. if value not found while comming out the loop return -1

```python
def binary_search(l1,val):
    start = 0
    end = len(l1) - 1

    while start <= end :
        mid = (start + end) // 2

        if l1[mid] == val:
            return f"found at {mid}"
        elif val < l1[mid]:
            end = mid - 1
        elif val > l1[mid]:
            start = mid + 1

    return -1


l1 = [8,10,14,22,27,36,55]
val = int(input("enter the value to search: "))
print(binary_search(l1, val))
```
