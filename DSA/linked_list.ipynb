# Linked List
   
  - it is a linear data structure created with the help of nodes and the links 
  - in this it can handle both homogenious as well as heterogenious including duplicate elements


### Node

- it is a memory alocation or memory block which contains two fields data field and address field
- data field is used to store the data given by the user and address field holds the address of its next node


                            node's address field 
                        /    
        |-------|---------| 
        |  DATA | Address |
        |-------|---------|
            /          address
        /               |                
        data field         |--- node'S address


    => syntax:

        class Node:
            def __init__(self, data):
                self.data = data
                self.address = None

        newnode = Node(10)

- when the object is created in the node class data fields contains the data and the address field will be None
- head node should be created in the begining and initially it is none and it always holds the address of first node

- note:
    - newnode           -> address of the node in the memory location
    - newnode.data      -> data inside the node's data field
    - newnode.address   -> address of the next node inised the node's address field

2. Display

        1. check head is none or not, if it is none display a messge
        2. if it is not none travverse till the end of linked list with the help of temp node

3. length

    2. if it is not none in the else part traverse till the end of single linked list using temp variable
    3. in each traversing count the node with count varible and atlast return the count

4. insert at first

    2. check if head is none assing the new nodes address to head 
    3. if it is not none in the else part assign first nods address present in head to new nodes address field
    4. then assign new nodes address to head

5. insert at the specified location

        7. if non of the condition is true in the else part traverse till the specified location's previous location with the help of temp and the node numbers
        8. when iteration is stoped insert new node at the specified location by moving current node to its next position

        2. in the else part traverse using temp variable
