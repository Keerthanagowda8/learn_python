# recurssion
- "it is a way of function calling itself"
- recurssion always happens in stack memeory
- when the function is called the function call is pushed to the stack memory


#### rules:
    1. Need to Pass smaller input or higher input in each function call to make problem smaller
    2. in each func call need pass different input
    3. Basic condition is required to stop the recursion 

```python
def facrorial(n):
    if n==1:
        return 1
    else:
        return n * facrorial(n-1)
```

### program to find sum of first 10 natural numbers

```python
def sum_of_nums(n):

    if n == 10:
        return 10
    else:
        return n + sum_of_nums(n+1)

print(sum_of_nums(1))
```

### WAP to find sum of a given number

```python
def sum_of_digit(n):

    if n == 0:
        return 0
    else:
        return n % 10 + sum_of_digit(n//10)
    

sum_of_digit(8234)
```

### program to convert decimal to binary using recurssion

```python
def decimal_to_binary(n):
    if n > 1:
        decimal_to_binary(n//2)
    print(n%2,end="")
    
decimal_to_binary(10)
```

### program using recursion to find the sum and product of elements present in a list

```python
def sum_of_list(l1):
    if len(l1) == 0:
        return 0
    else:
        return l1[0] + sum_of_list(l1[1:])
    
def product_of_list(l1):
    if len(l1) == 0:
        return 1
    else:
        return l1[0] * product_of_list(l1[1:])
    
print("sum of the list")
print(sum_of_list([1,2,3,4,5]))
print("product of the list")
print(product_of_list([1,2,3,4,5]))
```

### reverse the given string using recurssion

```python
def reverse_str(s1):
    if len(s1) == 1:
        return s1
    else:
        return s1[-1] + reverse_str(s1[:-1])
    
s1 = "python"


print(reverse_str(s1))
```

### program to check the given string is palindrom number or not

```python
def palindrom(s1):
    if len(s1)==0:
        return "palindrom"
    elif s1[0] != s1[-1]:
        return "not palindrome"
    else:
        return palindrom(s1[1:-1])
    
print(palindrom("madam"))
```

### print even numbers from 1 to 10

```python
def even_num(n):

    if n <= 10:
        print(n,end=" ")
        even_num(n+2)


even_num(2)
```

### convert each word to title case which is present in the list using recurssion

```python
data = ['python','coding','is','awesome','using','recurssion']

def solution(data):
    result = []
    if len(data) == 0:
        return result

    result.append(data[0].title())
    return result + solution(data[1:])
    
print(solution(data))
```
