### Operations can perform on stacks:

1. isFull()
    - it returns true is the stack is reached its limited size or returns false
2. isEmpty()
    - it returns true if the stack contains any empty space or returns False
2. push()
    - it is used to insert the element to the stack at the top end
4. pop()
    - it is used to remove the element from the stack at top end
5. display()
    - it is used to display all the elements in a stack
6. peek()
    - used to display topmost element in a stack

```python
class STACK:
    def __init__(self):
        self.s1 = []
        
    def display(self):
        if len(self.s1) == 0:
            print("stack is empty")
        else:
            print(self.s1)

    def push(self, value):
        self.s1.append(value)

    def peak(self):
        print(self.s1[-1])

    def pop(self):
        self.s1.pop()
    
st = STACK()
st.display()
st.push(10)
st.display()
st.push(20)
st.display()
st.peak()

print()
st.pop()
st.peak()
st.display()

```

## implementation of stack usign limited size

```python
class STACK:
    def __init__(self,maxSize: int):
        self.s1 = []
        self.maxSize = maxSize

    def isFull(self):
        if len(self.s1) == self.maxSize:
            return True
        else:
            return False
        
    def isEmpty(self):
        if len(self.s1) == 0:
            return True
        else:
            return False
        
    def display(self):
        if self.isEmpty():
            print("stack is empty")
        else: 
            for i in range(len(self.s1)-1, -1,-1):
                print(self.s1[i])

    def push(self,val):
        if self.isFull():
            print("Stack is full")
        else:
            self.s1.append(val)

    def pop(self):
        if self.isEmpty():
            print("stack is Empty")
        else:
            self.s1.pop()

    def peek(self):
        if self.isEmpty():
            print("Stack is Empty")
        else:
            print("top most element is, ", self.s1[-1])



st = STACK(5)

p = True
while p:
    print("***************** STACK OPERATIONS *******************")

    print("1. push\n 2. pop \n 3. peek\n 4. display\n 5. isFull\n 6. isEmpty\n 7. exist\n")

    c = int(input("Enter your choice: "))


    match c:
        case 1:
            val = int(input("Enter the val: "))
            st.push(val)
        
        case 2:
            st.pop()
        case 3:
            st.peek()
        case 4:
            st.display()
        case 5:
            print("stack is full: ", st.isFull())
        case 6:
            print("stack is empty: ", st.isEmpty())
        case 7: 
            p=False
```
