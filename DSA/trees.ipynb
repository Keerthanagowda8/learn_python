### tree terminolody

Node
 - it is a top most node in a tree
Links
 - it is a connection between parent node and child node
sibling nodes
 - it is a node which has same pair
leaf nodes
 - it is a nodes which dosen't have any children

 eg:

                     15
                   /    \
                 10      20
                /  \    /   \ 
               7   14  19    28
             /  \
            4   9 

- during insertion check root node is present in the tree or not
- if it is not present first value becomes the root node
- check if the root node is present new value is greater than the root or less than the root 
- if new value is greater than insert it into right side of the root, if it is less than insert into the left side of root node
- this proccess should happens recursively

### Operations of Trees 

1. Insertion
2. display
3. height of a tree
4. search the element
5. depth first search
    1. pre order
    2. in order
    3. post order
    
6. breadth first search
    1. level order
    2. zig-zag order


```python
class TREES:
    def __init__(self, data):  # we are not creating seperate class for node. using init we can create root node.
        self.prev = None
        self.data = data
        self.next = None

    def insert_at_ele(self,val):

        if self.data:   # check root node is present or not
            if val < self.data:  # check new value is lessthan the root node
                if self.prev is None: # if root node prev is none or not
                    self.prev = TREES(val) # insert new value to left side
                else:
                    self.prev.insert_at_ele(val)  # call recursivly the insert method only left side

            elif val > self.data: # check new valuse is greater than root node
                if self.next is None: # if root node next is none or not
                    self.next = TREES(val)  # insert the new value to right side
                else:
                    self.next.insert_at_ele(val) # call recursivly the insert method only right side


    # depth first search
    def pre_order_traversal(self,root):
        if root:
            print(root.data, end="-> ")     # start from root node
            self.pre_order_traversal(root.prev) # all left node
            self.pre_order_traversal(root.next) # all right node

    def in_order_traversal(self,root):
        if root:
            self.in_order_traversal(root.prev) # deepest left node
            print(root.data, end="-> ")         # root node
            self.in_order_traversal(root.next)  # right node

    def post_order_traversal(self,root):
        if root:
            self.post_order_traversal(root.prev)
            self.post_order_traversal(root.next)
            print(root.data, end="-> ")

    # breadth first search
    def search_ele(self,val):
        if val < self.data:
            if self.prev is None:
                return "value not found"
            self.prev.search_ele(val)
        elif val > self.data:
            if self.next is None:
                return "value not found"
            self.next.search_ele(val)

        else:
            return "value found"


t1 = TREES(15)
t1.insert_at_ele(20)
t1.insert_at_ele(10)
t1.insert_at_ele(14)
t1.insert_at_ele(7)
t1.insert_at_ele(9)
t1.insert_at_ele(4)
t1.insert_at_ele(20)
t1.insert_at_ele(19)
t1.insert_at_ele(28)

print("pre order traversal: ")
t1.pre_order_traversal(t1)
print()
print("in order traversal: ")
t1.in_order_traversal(t1)
print()
print("post order traversal: ")
t1.post_order_traversal(t1)
print()
```

# Depth First Search
1. pre order

    - traversing need to start form root node
    - travese to the left node of current root node, need to traverse till the deepest left root
    - once left node traveseing is completed go to the right node

               root
                |
              left
                |
              right

                n1->n2->n4->n8->n9->n5->n3->n6->n7

                     15

                 10      20
                /  \    /   \ 
               7   14  19    28
             /  \
            4   9 

            15-> 10-> 7-> 4-> 9-> 14-> 20-> 19-> 28-> 

2. in order

    - start traversing form root node and go till the last deepest left node
    - once reach with the deepest left node then go to its root node then travel to its right node

              left
                |
              root
                |
              right

                 10      20
                /  \    /   \ 
               7   14  19    28

            4   9 

              4-> 7-> 9-> 10-> 14-> 15-> 19-> 20-> 28-> 

3. post order

    - in this start traversing from root node
    - traverse till the last deepest left node of current root node
    - once left node traversing is completed go to the right node and then go to its root node

# Breadth first search

- in this start traversing form the root node
- need to traverse based on the number of levels present in a tree
- traverse all the nodes present in each level then go to its next level


                     15             -> level 1
                   /    \
                 10      20         -> level 2                        
                /  \    /   \ 
               7   14  19    28     -> level 3    
             /  \
            4   9                   -> level 4
