# Greedy Algorithm

```python
def coin_change(coins,amount):
    N = amount
    coins.sort()
    index = len(coins) - 1
    count = 0
    d={}
    while True:
        coinValue = coins[index]

        if N >= coinValue:
            print(coinValue)
            count += 1
            N = N - coinValue

        if N < coinValue:
            d[coinValue] = count
            count = 0
            index = index - 1

        if N == 0:
            print("division: ", d)
            print("total count of coins: ", count)
            break

coins = [1, 2, 5, 10,20, 50, 100, 200, 500, 1000]
amount = int(input("enter the total amount: "))
coin_change(coins,amount)
```
